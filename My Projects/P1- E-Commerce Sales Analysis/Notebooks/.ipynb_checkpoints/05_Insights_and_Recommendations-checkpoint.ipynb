# Project 1: E-Commerce Sales Analysis
## Phase 6 — Business Insights & Recommendations

### 1. Objective

The objective of this phase is to convert the findings from Exploratory Data Analysis into actionable business recommendations. The focus is on improving profitability, identifying underperforming products and regions, and finding opportunities for sustainable growth.

### 2. Key Business Insights and Recommendations

#### Insight 1: Furniture has a significant profitability problem

**Finding:** Furniture generated sales of 754,747.76 but profit of only 19,730.00, giving it a profit margin of 2.61%. Technology and Office Supplies achieved margins of 17.45% and 17.22%, respectively.

**Recommendation:** Investigate Furniture products with low or negative profit, review discount levels, and assess whether pricing or product costs are affecting profitability. Prioritize Tables and other loss-making products for further analysis.

**Success metric:** Furniture profit margin and total profit.

#### Insight 2: Technology is the strongest product category

**Finding:** Technology generated the highest category sales of 839,893.28 and the highest profit of 146,543.38.

**Recommendation:** Examine the products and customer groups contributing most to Technology's profitability. Use these findings to identify opportunities for growth while protecting existing margins.

**Success metric:** Technology sales, profit, and profit margin.

#### Insight 3: The West region leads business performance

**Finding:** West generated the highest regional sales of 739,813.61 and profit of 110,798.82. Central generated more sales than South but less profit, with a lower profit margin.

**Recommendation:** Study the product mix and discount patterns in Central to identify why its profitability is weaker. Compare these patterns with West and South before deciding on corrective actions.

**Success metric:** Regional profit margin and total profit.

#### Insight 4: Consumer is the largest customer segment, but Home Office has the highest margin

**Finding:** Consumer generated sales of 1,170,659.79 and profit of 136,371.45. However, Home Office achieved the highest segment profit margin at 14.02%, compared with Consumer's 11.65%.

**Recommendation:** Maintain the Consumer segment as a key business priority while investigating opportunities to grow Home Office sales without reducing its profit margin.

**Success metric:** Sales, profit, and profit margin by customer segment.

#### Insight 5: Product sales and profitability are not always aligned

**Finding:** Chairs generated the highest sub-category sales at 335,768.25, whereas Copiers generated the highest sub-category profit at 56,093.94. Tables recorded a loss of 17,753.21.

**Recommendation:** Evaluate products using both sales and profit rather than sales alone. Investigate Tables, Bookcases, and Supplies for potential profitability issues, and assess the factors contributing to Copiers' strong performance.

**Success metric:** Product-level profit, profit margin, and loss-making product count.

#### Insight 6: Higher discount levels are associated with weaker profitability

**Finding:** The discount analysis showed negative total profit at several discount levels of 30% and above. The 70% discount level recorded a combined loss of 40,300.57.

**Recommendation:** Review heavily discounted transactions and identify the products and categories responsible for the losses. Consider introducing stricter discount approval rules where discounts consistently undermine profitability. Validate product mix and transaction characteristics before changing the discount policy.

**Success metric:** Profit margin by discount level and total profit from discounted transactions.

#### Insight 7: Office Supplies has many loss-making records

**Finding:** Office Supplies had 898 loss-making records, the highest count among the three categories. Furniture recorded the largest combined category loss at 61,675.18.

**Recommendation:** Prioritize Office Supplies for investigating the frequency of loss-making records and Furniture for investigating the total financial impact. Review the underlying products and transactions to determine where corrective action may be useful.

**Success metric:** Number of loss-making records and total losses by category.

### 3. Priority Action Plan

| Priority | Recommended action | Expected business benefit |
|---|---|---|
| 1 — High | Investigate Furniture profitability and loss-making products | Identify opportunities to improve margins |
| 2 — High | Review heavily discounted transactions | Reduce potential discount-related losses |
| 3 — High | Investigate Central region profitability | Identify regional performance gaps |
| 4 — Medium | Evaluate high-profit products such as Copiers and Phones | Identify sustainable growth opportunities |
| 5 — Medium | Explore Home Office segment growth | Increase sales while protecting margins |
| 6 — Medium | Review loss-making Office Supplies products | Reduce the frequency of unprofitable transactions |

These priorities are recommendations based on the observed patterns. Their actual impact must be measured after the business implements any changes.

### 4. Limitations

The available analysis identifies patterns but does not establish the causes of low profitability. Discounting, product mix, pricing, and cost factors require further investigation before specific operational changes can be justified.

The dataset also lacks an explicit product-return indicator and customer age field, so return-rate and age-based recommendations cannot be supported by this project.

The reported date range extends into December 2026 and must be validated before the results are used for time-based business decisions.

### 5. Conclusion

The analysis highlights Furniture profitability, discount practices, and Central region performance as important areas for investigation. Technology is the leading category by total sales and profit, West is the strongest region, and Consumer is the largest customer segment. At the same time, Home Office has the highest segment profit margin, and Copiers is the most profitable sub-category by total profit.

The recommended approach is to prioritize loss prevention and profitability analysis while exploring growth opportunities in stronger products and customer segments. These insights provide a business-focused foundation for the next phase: building an interactive Power BI dashboard to communicate performance, highlight problem areas, and support decision-making.
