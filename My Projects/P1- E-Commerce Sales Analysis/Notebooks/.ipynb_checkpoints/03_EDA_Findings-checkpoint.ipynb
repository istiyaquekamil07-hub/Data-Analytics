# Project 1: E-Commerce Sales Analysis

## Exploratory Data Analysis (EDA) Report

### 1. Objective of EDA

The objective of Exploratory Data Analysis was to understand the business's overall performance, identify its most profitable and least profitable product categories, evaluate regional and customer segment performance, investigate the relationship between discounts and profit, and examine sales and profit trends over time.

The analysis used Python, Pandas, Matplotlib, and Seaborn to summarize the dataset, calculate business metrics, identify patterns, and visualize relationships between variables.

### 2. Dataset Overview

The dataset contains **10,194 records and 21 columns**, covering order information, customer details, product categories, geographical regions, sales, quantities, discounts, and profit.

The analysis identified:

- 5,111 unique orders
- 804 unique customers
- 1,862 unique products
- 3 product categories
- 17 product sub-categories
- 4 geographical regions
- 3 customer segments
- 4 shipping modes

The dataset contains numerical variables such as Sales, Quantity, Discount, and Profit, alongside categorical variables such as Category, Region, Segment, and Sub-Category.

The reported order dates span January 2023 through December 2026. This date range requires validation because it includes dates later than the current analysis date.

### 3. Overall Business Performance

The overall business metrics are summarized below.

| Metric                |       Result |
| --------------------- | -----------: |
| Total Sales           | 2,326,534.35 |
| Total Profit          |   292,296.81 |
| Overall Profit Margin |       12.56% |
| Total Quantity Sold   |       38,654 |
| Unique Orders         |        5,111 |
| Unique Customers      |          804 |
| Unique Products       |        1,862 |

The business generated approximately 2.33 million in sales and 292,296.81 in profit, with an overall profit margin of 12.56%.

The average sales value per record was 228.23, while the median was 53.91. This difference indicates that sales values are right-skewed: many records have relatively small sales values, while a smaller number of records have much larger values.

Sales ranged from 0.44 to 22,638.48. Profit ranged from a loss of 6,599.98 to a profit of 8,399.98 per record. These extremes highlight the need to investigate unusually large transactions and significant losses without automatically treating them as data errors.

### 4. Product Category Performance

Technology generated the highest sales and total profit among the three categories.

| Category        |      Sales |     Profit | Profit Margin |
| --------------- | ---------: | ---------: | ------------: |
| Technology      | 839,893.28 | 146,543.38 |        17.45% |
| Furniture       | 754,747.76 |  19,730.00 |         2.61% |
| Office Supplies | 731,893.31 | 126,023.44 |        17.22% |

**Key findings:**

- Technology was the strongest category by total sales and profit, with a profit margin of 17.45%.
- Office Supplies generated lower sales than Furniture but substantially higher profit. Its profit margin of 17.22% was close to Technology's.
- Furniture generated considerable sales but achieved a profit margin of only 2.61%, making it the most significant category-level profitability concern.

**Business implication:** Furniture should be investigated to understand whether discounting, product mix, or other factors are contributing to its low profitability. Technology and Office Supplies provide stronger profitability benchmarks for further analysis.

### 5. Regional Performance

Regional analysis revealed differences between sales volume and profitability.

| Region  |      Sales |     Profit | Profit Margin |
| ------- | ---------: | ---------: | ------------: |
| West    | 739,813.61 | 110,798.82 |        14.98% |
| East    | 691,828.17 |  94,883.26 |        13.71% |
| Central | 503,170.67 |  39,865.31 |         7.92% |
| South   | 391,721.91 |  46,749.43 |        11.93% |

**Key findings:**

- West ranked first in both sales and total profit, with the highest regional profit margin of 14.98%.
- East was the second-largest region by sales and profit.
- Central generated more sales than South but lower total profit and a lower profit margin.
- South recorded the lowest sales but generated more profit than Central.

**Business implication:** West represents the strongest region in the reported results. Central warrants further investigation because its profitability is relatively weak despite higher sales than South. Product mix and discount practices are potential areas to examine, but the current analysis does not establish the cause.

### 6. Customer Segment Performance

The analysis compared Consumer, Corporate, and Home Office customers.

| Segment     |        Sales |     Profit | Profit Margin |
| ----------- | -----------: | ---------: | ------------: |
| Consumer    | 1,170,659.79 | 136,371.45 |        11.65% |
| Corporate   |   715,806.13 |  94,249.64 |        13.17% |
| Home Office |   440,068.43 |  61,675.73 |        14.02% |

**Key findings:**

- Consumer generated the highest sales and total profit, with 2,628 unique orders.
- Corporate ranked second in sales and total profit.
- Home Office generated the lowest sales and total profit but achieved the highest profit margin at 14.02%.
- Consumer had the lowest profit margin among the three segments, at 11.65%.

**Business implication:** Consumer is the largest contributor to overall business volume and profit. Home Office appears more profitable relative to its sales, suggesting an opportunity to investigate ways to grow this segment while maintaining its margin.

### 8. Discount Analysis

The discount analysis examined how sales and profit varied across the discount levels recorded in the dataset.

| Discount |        Sales |     Profit | Profit Margin |
| -------- | -----------: | ---------: | ------------: |
| 0%       | 1,105,323.79 | 326,718.59 |        29.56% |
| 10%      |    54,952.50 |   9,099.97 |        16.56% |
| 15%      |    27,558.52 |   1,418.99 |         5.15% |
| 20%      |   773,939.40 |  91,079.95 |        11.77% |
| 30%      |   104,474.07 | -10,513.45 |       -10.06% |
| 32%      |    14,493.46 |  -2,391.14 |       -16.50% |
| 40%      |   116,497.76 | -23,086.37 |       -19.82% |
| 45%      |     5,484.97 |  -2,493.11 |       -45.45% |
| 50%      |    58,918.54 | -20,506.43 |       -34.80% |
| 60%      |     7,105.94 |  -6,164.35 |       -86.75% |
| 70%      |    40,805.27 | -40,300.57 |       -98.76% |
| 80%      |    16,980.15 | -30,565.27 |      -180.01% |

**Key findings:**

- Records without discounts generated the highest reported sales and a profit margin of 29.56%.
- The 20% discount level accounted for substantial sales but had a lower profit margin of 11.77%.
- Discount levels of 30% and above recorded negative total profit in the reported results.
- The 70% discount level generated a loss of 40,300.57, while the 80% level generated a loss of 30,565.27.
- Profit margins became increasingly negative at several higher discount levels.

**Business implication:** The results show a strong association between higher recorded discount levels and weaker profitability. The business should investigate its discounting strategy, particularly for heavily discounted transactions.

However, these results do not prove that discounts alone caused the losses. Product mix, transaction characteristics, and the number of records at each discount level may also influence the results. The dataset should be examined at the product and transaction levels before making pricing decisions.

### 9. Correlation Analysis

The correlation matrix examined the linear relationships between Sales, Quantity, Discount, and Profit.

| Variable Pair         | Correlation |
| --------------------- | ----------: |
| Sales and Profit      |        0.48 |
| Discount and Profit   |       -0.22 |
| Quantity and Sales    |        0.20 |
| Quantity and Profit   |        0.07 |
| Sales and Discount    |       -0.03 |
| Quantity and Discount |        0.01 |

**Key findings:**

- Sales and Profit had a moderate positive correlation of 0.48. Higher sales values tended to be associated with higher profit values, although this relationship was not perfect.
- Discount and Profit had a negative correlation of -0.22, consistent with the observed pattern of weaker profitability at higher discount levels.
- Quantity and Sales had a weak positive correlation of 0.20.
- Quantity and Profit had a very weak positive correlation of 0.07.
- Sales and Discount had a correlation close to zero at -0.03.

**Business implication:** Higher sales values tend to accompany higher profits, but selling more units does not necessarily result in substantially higher profit. The negative relationship between discount and profit supports further investigation into discount practices. Correlation measures association, not causation.

### 10. Monthly Sales and Profit Trends

Monthly analysis aggregated Sales, Profit, Quantity, and unique Orders by order month.

The reported monthly results show noticeable variation in sales and profit across the period. Sales and order counts are relatively high in several later months, while some months have substantially lower values. Profit also fluctuates, and at least two months in the displayed results—July 2023 and January 2024—recorded negative monthly profit.

The monthly table reports the following notable observations:

- September 2023 generated sales of 82,670.43 and profit of 8,593.63.
- November 2024 generated profit of 12,474.79.
- December 2025 generated sales of 97,502.28 and profit of 17,926.30.
- November 2026 is reported as the highest-sales month in the provided monthly table, at 118,454.51.

**Important limitation:** The dataset includes monthly records through December 2026, including months that have not yet occurred as of the analysis date. Therefore, the latest monthly results cannot be treated as verified historical performance. The date range and source data must be checked before drawing final conclusions about growth, seasonality, or the best-performing month.

### 11. Loss-Making Records

The analysis identified **1,901 records with negative profit**.

| Category        | Loss-Making Records | Combined Loss |
| --------------- | ------------------: | ------------: |
| Furniture       |                 732 |    -61,675.18 |
| Office Supplies |                 898 |    -56,783.83 |
| Technology      |                 271 |    -38,579.92 |

Office Supplies had the highest number of loss-making records, while Furniture had the largest combined loss.

These counts refer to individual dataset records with negative profit, not necessarily unique orders or customers. An order can contain multiple product records, and multiple records from the same order may be loss-making.

**Business implication:** Furniture and Office Supplies should be prioritized for further profitability investigation. Reviewing heavily discounted products, loss-making sub-categories, and individual transactions may help identify specific areas requiring attention.

### 12. Data Limitations

The following limitations should be considered when interpreting the EDA findings:

1. **Date validity:** The reported dataset extends through December 2026. Dates later than the current analysis date require verification before monthly or yearly conclusions are finalized.
2. **No return indicator:** The dataset does not contain an explicit return field, so product return rates cannot be calculated.
3. **No customer age:** Customer age-group purchasing patterns cannot be evaluated from the available columns.
4. **Repeated order IDs:** An order can contain multiple product records. Therefore, unique order counts calculated within different groups may overlap and should not be added together.
5. **Loss-making records:** Negative-profit record counts are not equivalent to unique loss-making orders or customers.
6. **Outliers:** Sales and Profit contain extreme values. These should be investigated for business validity rather than automatically removed.
7. **Correlation and causation:** Observed relationships between discounts, sales, and profit do not independently establish causation.

### 13. Final EDA Conclusion

The EDA identified several important patterns in the reported dataset. Technology is the strongest product category by total sales and profit, while West is the strongest region by sales, profit, and profit margin. Consumer is the largest customer segment by sales and total profit, whereas Home Office has the highest segment profit margin.

At the sub-category level, Chairs generates the highest sales, while Copiers generates the highest profit. Tables records the largest sub-category loss. At the category level, Furniture stands out as the main profitability concern because its profit margin is only 2.61% despite substantial sales. The discount analysis also shows that several higher discount levels are associated with negative profit, and the correlation analysis shows a negative relationship between discount and profit.

These findings suggest that future analysis should prioritize Furniture profitability, loss-making products, discounting practices, and the comparatively weak performance of the Central region. Opportunities may also exist to grow the Home Office segment while protecting its relatively high profit margin.

**EDA status:** The exploratory analysis has produced a useful set of descriptive findings and candidate business priorities. Before the project is considered fully validated, the reported future dates should be checked against the source data. The next project phase is **SQL Analysis**, where the business questions and reported metrics will be examined through SQL queries.
