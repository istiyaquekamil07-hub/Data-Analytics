### Step 3 - Data Transformation

```python
# Converting data types safely
df["Order Date"] = pd.to_datetime(df["Order Date"], format="mixed", errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="mixed", errors="coerce")
```

```python
df[["Sales", "Profit", "Discount", "Quantity"]].describe()
```

```python
#Standardize Columns Names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" " , "_")
    .str.replace("/" , "_")
)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Checking Missing Value
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Checking Duplicate Rows

duplicate_rows = df.duplicated().sum()
print("\nDuplicate complete rows:", duplicate_rows)
```

```python
 # Validating important values

print("\nNegative Sales:", (df["sales"] <0).sum())
print("\nNegative Profit:", (df["profit"]<0).sum())
print("\nNegative Quantity:", (df["quantity"]<= 0).sum())
print("\nInvalid discounts:", (df["discount"]<0) | (df["discount"] > 1).sum())
```
