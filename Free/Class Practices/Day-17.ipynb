Importing image chaging its colour

```python
# Importing Requried Libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
x= r'image/Shah-Rukh-Khan.jpg'

#opening image using pil

image= Image.open(x).covert('L')

# Mapping image to grayscale
plt.imshow(image, cmap='gray')
plt.show()
```

Pie Chart

```python
x=[10,20,30,40]
y=['english','math','bengali','hindi']

plt.pie(x, labels=y) #showcasing subject name
plt.show()
```

Changing Colours in Pie chart

```python
x=[10,20,30,40]
y=['english','math','bengali','hindi']
c=['yellow','magenta','red','aqua'] #for changing color
plt.pie(x, labels=y, colors=c) #showcasing subject name
plt.show()
```

Creating Legend Lables in Pie chart

```python
x=[10,20,30,40]
y=['english','math','bengali','hindi']

plt.pie(x, labels=y) #showcasing subject name
plt.legend()  # creating legends
plt.show()
```
