BAR Graph

```python
x=[1,2,3,4,5]
y=[5,7,10,1,6]
plt.bar(x,y)   #creatig bar graph
plt.show()
```

Adding Colour in Bars


```python
x=[1,2,3,4,5]
y=[5,7,10,1,6]
c=['r','g','b','y']  # Making Color Varible
plt.bar(x,y,color=c)    # Adding Color  
plt.show()
```

Horizental Bar

```python
import numpy as np

x=np.array(["A","B","C","D"])
y=np.array([3,8,1,10])

plt.barh(x,y)
plt.show()
```

```python
x=np.array(["A","B","C","D"])
y=np.array([3,8,1,10])

plt.bar(x,y,color="red")
plt.show()
```

Minimizing the width

```python
x=np.array(["A","B","C","D"])
y=np.array([5,8,10,13])
plt.bar(x,y,width=0.2)      #width minimization
plt.show()
```

Minimizing bar in horizental graph

```python
x=np.array(["A","B","C","D"])
y=np.array([5,8,10,13])
plt.barh(x,y,height=0.1)      #width minimization
plt.show()
```
