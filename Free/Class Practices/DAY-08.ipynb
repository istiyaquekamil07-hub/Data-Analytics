### Change Range of Items

```python
a=[12,15,23,34,54]
a[1:5]=["nook","pen","paper"]
print(a)
```

```python
a=[12,15,23,34,54]
a.append(-5)
print(a)
```

```python
a=[12,15,23,34,54]
a.remove(12)
print(a)
```

### Reverse Item
