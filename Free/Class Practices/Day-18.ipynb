Using dictionary dataframe and making graph from Seaborn 

```python
var=[1,2,3,4,5,6,7,8]
var1=[1,7,4,5,3,2,1,6]
import pandas as pd

z=pd.DataFrame({"var":var,"var1":var1})
sns.lineplot(x='var',y='var1',data=z)
plt.show()
```

Importing dataset from github

```python
import seaborn as sns
dg=sns.load_dataset("penguins")   #importing dataset
dg
```

Making graph from seaborn 

```python
import matplotlib.pyplot as plt
sns.lineplot(x="bill_length_mm",y="flipper_length_mm", data=dg)
plt.show()
```

Categorized via Hue 

```python
dg
sns.lineplot(x="bill_length_mm",y="flipper_length_mm", data=dg,hue="sex")
plt.show()
```

Colour Acc to me



```python
dg
sns.lineplot(x="bill_length_mm",y="flipper_length_mm", data=dg,hue="sex",palette="rocket_r")
plt.show()
```

Adding Marker

```python
dg
sns.lineplot(x="bill_length_mm",y="flipper_length_mm", data=dg,hue="sex",palette="rocket_r", markers=["o",">"], style="sex")
plt.show()
```

Adding Grid

```python
dg
sns.lineplot(x="bill_length_mm",y="flipper_length_mm", data=dg,hue="sex",palette="rocket_r", markers=["o",">"], style="sex")
plt.show()
plt.grid()
```

Adding Title

```python
dg
sns.lineplot(x="bill_length_mm",y="flipper_length_mm", data=dg,hue="sex",palette="rocket_r", markers=["o",">"], style="sex")

plt.grid()
plt.title("Penguins Graph")
plt.show()
```
