### Seaborn Library 2

Bar Graph in Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ds=sns.load_dataset("penguins")
```

```python
ds
```

```python
ds["island"]
```

```python
sns.barplot(x=ds.island, y=ds.bill_length_mm)
```

```python
sns.barplot(x='island',y='bill_length_mm',data=ds)
```

```python
ds
```

```python
sns.barplot(x='island',y='bill_length_mm',hue='sex',data=ds)
```

Order Parameter

```python
od=["Dream","Torgersen","Biscoe"]
sns.barplot(x='island',y='bill_length_mm',data=ds,order=od)
```

Changing the order of bars sepration


```python
od=["Dream","Torgersen","Biscoe"]
sns.barplot(x='island',y='bill_length_mm',data=ds,order=od,hue='sex',hue_order=['Female','Male'])
```

Changing Colours of Bars

```python

sns.barplot(x='island',y='bill_length_mm',data=ds,color='green')
```

```python
sns.barplot(x='island',y='bill_length_mm',data=ds,palette="prism",hue="sex")
```

Saturation Parameter

```python
sns.barplot(x='island',y='bill_length_mm',data=ds,palette="prism",saturation=0.5,hue="sex")
```

### Histogram in seaborn library

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ds
```

```python
sns.displot(ds["flipper_length_mm"])
plt.show()
```

Bins Parameter
used for proper alignment between bars

```python
sns.displot(ds["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240])
```

color parameter

```python
sns.displot(ds["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240],color="green")
```

Kernel Density (KDE)

```python
sns.displot(ds["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240],kde=True,color="red")
```
