Sets is Python

Set Constructor

```python
x=set(("Alpha","Beta","Gama"))
print(x)
print(type(x))
```

Empty Set

```python
x=set()
print(x)
```

Check item 
###### if item is present in code then it will show true and if not the false

```python
x={"data","python","sql"}
print("sql"in x)
```

Add item

```python
x={"a","b","c"}
x.add(100)
print(x)
```

Combine item

```python
x={"boy","girl"}
y={10,20}
x.update(y)
print(x)
```

Remove item

```python
x={"boy","girl"}
x.remove("girl")
print(x)
```

```python
x={"boy","girl"}
x.pop()   #with this you can remove randomly
print(x)
```

```python
x={"boy","girl"}
x.clear()   # this can clear whole set in a single comand
print(x)
```

Join Two Set

```python
x={"a","b","c"}
y={1,2,3}
z=y.union(x)
print(z)
```

Find Max and Min Value

```python
x={5,45,43,22,33,65}
print("Orignal set elements")
print(x)
print("Max value of set is")
print(max(x))
print("min value of set is")
print(min(x))
```

Find Identical Values

```python
x={"boy","girl"}
y={10,30,"boy"}
print(x.intersection(y))
```
