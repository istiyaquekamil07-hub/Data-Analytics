Creation of Array

```python
import numpy as np
```

```python
x=np.array([1,2,3,4])
print(x)
print(type(x))
```

Checking Dimension of Array

```python
x=np.array([1,2,3,4])
print(x)
print(x.ndim)
```

1D Array

```python
x=np.array([1,2,3,4])
print(x)
print(x.ndim)
```

2D Array

```python
x=np.array([[1,2,3],[4,5,6]])
print(x)
print("Dimension of array:",x.ndim)
```

3D Array

```python
x=np.array([[[1,2,3],[4,5,6],[7,4,3]]])
print(x)
print(x.ndim)
```

High Dimension Array

```python
x=np.array([1,2,3,4,5,6,33,22,65,42],ndmin=5)
print(x)
print("Number of Dimension:",x.ndim)
```
