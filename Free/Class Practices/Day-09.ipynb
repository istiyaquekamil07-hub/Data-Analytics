#### ALL ABOUT TUPLES
Python Tuple is a collection of objects separated by commas. In some ways, a tuple is similar to a Python list in terms of indexing, nested objects, and repetition but the main difference between both is Python tuple is immutable, unlike the Python list which is mutable.

#### Creating Tuple

```python
variable=(1,2,3,"apple")
print(variable)
```

#### Nested Tuple
In Python, a tuple written inside another tuple is known as a nested tuple

```python
a=("data",[1,2,3],(4,5,6),("apple","mango","cherry"))
print(a)
print(type(a))
```

#### Creating Tuple with one item
must use one comma inside this type of tupple either it will not work

```python

a=("hello",)
print(a)
print(type(a))
```

### Tuple Constructor

```python
a=tuple(("a","b","c"))
print(a)
```

### Adding item in tuple

```python
a=(23,12,24,21)
y=list(a)         #converted in list to adding item
y.append(20)      #adding item
a=tuple(y)        #coverting in tuple
print(a)      
```

### Removing Item in Tuple

```python
a=(23,12,24,21)
y=list(a)         #converted in list to removing item
y.remove(21)      #removing item
a=tuple(y)        #coverting in tuple
print(a)   
```

### Looping of Tuple


```python
b=("a","b","c")
for l in b:                #Here l and b both are variables
    print(l)
```

### Joining of Tuple

```python
b=("a","b","c")
d=("aman","ben","den")
z=b+d
print(z)
```


### Multiplying item in tupple

```python
d=("aman","ben","den")
x=d*3
print(x)
```
