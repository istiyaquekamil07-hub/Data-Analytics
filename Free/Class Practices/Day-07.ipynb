### INDEXING OF STRINGS
#### IT ALLOW US TO ACESS INDIVIDUAL CHARACTER

```python
c="Hello universe"
print(c[9])
```

#### Acessing Backward Characters

```python
c="Hello universe"
print(c[-1])
```

### Slicing of Strings
##### it helps to slice out a range of words

```python
c="Hello universe"
print(c[6:9])
```

### Reverse String
##### it helps to rewrite the string in reverse manner

```python
c="Hello universe"
print(c[ : :-1])
```

### Length of strings
##### it help us to know the length of string

```python
c="Hello universe"
print(len(c))
```

### Upper Method
##### Turns on Cap letter


```python
c="Hello universe"
print(c.upper())
```

### Lower Method
##### Turns on small letter

```python
c="Hello universe"
print(c.lower())
```

### Replace Function


```python
c="Hello universe"
print(c.replace("Hello","Bye"))
```

### Find Method
##### Find Index position of any letter

```python
c="Hello universe"
d=c.find("u")
print(d)
```
